```python
import torch
import torch.nn as nn
```

```python
x = torch.randn(10,4)

layer = nn.Linear(4,3)
y = layer(x)
print(y.shape)
```

```python
x = torch.randn(10,5)

layer1 = nn.Linear(5,4)
layer2 = nn.Linear(4,3)
layer3 = nn.Linear(3,2)

h1 = layer1(x)
h2 = layer2(h1)
h3 = layer3(h2)

print("Shape of h1: ",h1.shape)
print("Shape of h2: ",h2.shape)
print("Shape of h3: ",h3.shape)
```

```python
layer1 = nn.Linear(2,4)
layer2 = nn.Linear(4,1)
activation = nn.ReLU()

x = torch.rand(10,2)

h = layer1(x)
h = activation(h)
y = layer2(h)

print('X shape: ', x.shape)
print('Y shape: ', y.shape)
print('H Shape: ', h.shape)
```
