```python
import torch

```

```python
x = torch.tensor([1.,2.,3.])
w = torch.tensor([1.,2.,3.])

element_wise = x * w
dot = torch.dot(x,w)

print("Element wise: ",element_wise)
print("Element wise shape: " , element_wise.shape)
print("Dot: ",dot)
print("Dot Shape: ",dot.shape)
```

```python
x = torch.tensor(2.0)
y = torch.tensor(10.0)

w = torch.nn.Parameter(torch.tensor(6.0))
optimizer = torch.optim.SGD([w],lr=0.01)

#make a prediction
prediction = w * x

# calculate the loss
loss = ( prediction - y ) ** 2

#backward pass
optimizer.zero_grad()
loss.backward()

print("Prediction: ", prediction.item())
print("Loss: ", loss.item())
print("Gradient: ", w.grad.item())
print("Weight before update: ", w.item())

#update the weight
optimizer.step()
print("Weight after update: ", w.item())
```
