```python
import torch

```

```python
#initializing the weights
w1 = torch.tensor(2.0, requires_grad=True)
w2 = torch.tensor(3.0, requires_grad=True)

x = torch.tensor(2.0)
target = torch.tensor(10.0)

```

```python
h1 = w1 * x
a1 = torch.relu(h1)
h2 = w2 * a1

loss = 0.5 * (target - h2) ** 2
```

```python
loss.backward()
```

```python
print("Loss: ",loss.item())
print("w1: ",w1.grad.item())
print("w2: ",w2.grad.item())
```

```python
learning_rate = 0.01

with torch.no_grad():
    w1 = w1 - learning_rate * w1.grad
    w2 = w2 - learning_rate * w2.grad
print("w1: ",w1.item())
print("w2: ",w2.item())
```

```python
x = torch.tensor([[2.0], [3.0]])
w = torch.tensor([[4.0], [5.0]], requires_grad=True)

y = ( w.T @ x ).squeeze()

y.backward()

print("w: ",w.grad)
print("output: ",y.item())
```
