```python
corpus = '''
Hi Im Chethan, i like to play games.
I usually play Story games like RDR2, Hogwarts Legacy and tomb raider series.
And I Dont play multiplayer games because i dont have friends to play with.
'''
```

```python
from nltk.tokenize import sent_tokenize, word_tokenize, wordpunct_tokenize, TreebankWordTokenizer
```

```python
sentence_tokens = sent_tokenize(corpus, language="english")
```

```python
sentence_tokens
```

```python
word_tokens = word_tokenize(corpus)
```

```python
word_tokens
```

```python
tokenizer = TreebankWordTokenizer()
wrod_tree_tokens = tokenizer.tokenize(corpus)
wrod_tree_tokens
```
