```python
import torch
```

```python
#training data
x = torch.tensor([1.,2.,3.,4.])
y = torch.tensor([2.,4.,6.,8.])

#one learnable parameter
w = torch.nn.Parameter(torch.tensor(0.0))

#optimizer
optimizer = torch.optim.SGD([w], lr=0.01)

for epoch in range(100):

    #1. predict
    prediction = w * x

    #2. calculate loss
    loss = ((prediction - y) ** 2).mean()

    #3. clear the old gradients
    optimizer.zero_grad()

    #4. calculate the new gradient
    loss.backward()

    #5. update the parameter
    optimizer.step()

print("learned w: ",w.item())
print("final loss: ", loss.item())



```

```python
#training data
x = torch.tensor([1.,2.,3.,4.])
y = torch.tensor([2.,4.,6.,8.])

#one learnable parameter
w = torch.nn.Parameter(torch.tensor(0.0))

#optimizer
optimizer = torch.optim.SGD([w], lr=0.001)

for epoch in range(100):

    #1. predict
    prediction = w * x

    #2. calculate loss
    loss = ((prediction - y) ** 2).mean()

    #3. clear the old gradients
    optimizer.zero_grad()

    #4. calculate the new gradient
    loss.backward()

    #5. update the parameter
    optimizer.step()
    print(epoch, w.item(), loss.item())

print("learned w: ",w.item())
print("final loss: ", loss.item())



```
