```python
#if the tensors on different devices like one on the cpu and another on the gpu they will fail if did some operation on them.
x = torch.rand(1000, device=torch.device('cuda'))
y = torch.randn(1000)

z = x + y #device mismatch error
```

```python
y = y.to(x.device)

z = x + y
```
